```python
import os
import numpy as np
from skimage.io import imread
from skimage.transform import resize
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
from sklearn import svm
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score, recall_score, precision_score
import pandas as pd
```

```python
# Create categories for classification
Categories = ['Tumorous', 'Non-Tumorous']
flat_array = []
target_array = []
data_dir = r'C:\Users\emily\OneDrive\Desktop\Grace Capstone'

# Loop to retrieve images from OS, then resize them and add to array. Confirm once finished
for i in Categories:
    for file in os.listdir(os.path.join(data_dir, i)):
        img_path = os.path.join(data_dir, i, file)
        img_array = imread(img_path)
        img_resized = resize(img_array, (150, 150, 3))
        flat_array.append(img_resized.flatten())
        target_array.append(Categories.index(i))
    print(f'Loaded and resized {i} images')


```

```python
# Create / store in Numpy arrays
X = np.asarray(flat_array)
y = np.asarray(target_array)
```

```python
# Train test split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=32, stratify=y, shuffle=True)
```

```python
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [0.0001, 0.001, 0.1, 1],
    'kernel': ['poly', 'rbf']
}

svc = svm.SVC(probability=True)
model = GridSearchCV(svc, param_grid)

model.fit(X_train, y_train)
```
